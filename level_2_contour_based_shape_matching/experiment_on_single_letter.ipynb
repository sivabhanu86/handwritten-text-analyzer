```python
import cv2 
import matplotlib.pyplot as plt
import pandas as pd
```

```python
templete = cv2.imread('templeteA.jpg')
```

```python
plt.imshow(templete)
```

```python
temp_gray = cv2.cvtColor(templete,cv2.COLOR_BGR2GRAY)
```

```python
plt.imshow(temp_gray,cmap='gray')
```

```python
_,temp_threshold = cv2.threshold(
    temp_gray,
    80,
    255,
    cv2.THRESH_BINARY
)
plt.imshow(temp_threshold,cmap = 'gray')
```

```python
temp_contours , hierarcy = cv2.findContours(
    cv2.bitwise_not(temp_threshold),
    cv2.RETR_EXTERNAL,
    cv2.CHAIN_APPROX_SIMPLE
)
```

```python
print(len(temp_contours))
```

```python
drawing_image = templete.copy()
cv2.drawContours(
    drawing_image,
    temp_contours,
    -1,
    (0,255,0),
    2
)
plt.imshow(drawing_image)
```
